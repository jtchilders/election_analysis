# 2026 Midterm Election Analysis

This notebook uses the standardized, year-agnostic common schema (`common.schemas`) and plotting suite (`common.plots`) to analyze the 2026 Midterm Elections (U.S. Senate and U.S. House of Representatives).

Data is extracted via `2026/US_Senate_results.py` and `2026/US_House_results.py` using `PoliteFetcher` for rate-limiting and local caching.

```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt

# Ensure common library is importable
REPO_ROOT = os.path.abspath('..')
if REPO_ROOT not in sys.path:
    sys.path.insert(0, REPO_ROOT)

from common.plots import plot_balance_of_power, plot_party_seats, plot_turnout_projections
%matplotlib inline
```

## 1. U.S. Senate Analysis (2026)

- Total Senate Seats: 100 (50 needed for majority)
- Baseline Seats Held (Not up for election in 2026): Democratic: 34, Republican: 31
- 35 Senate Contests in 2026 (Class 2 seats + special elections)

```python
senate_state_csv = 'US_Senate_state_results.csv'
senate_county_csv = 'US_Senate_county_results.csv'

if os.path.exists(senate_state_csv):
    df_senate_state = pd.read_csv(senate_state_csv)
    display(df_senate_state)
    plot_balance_of_power(df_senate_state, office='Senate', held_dem=34, held_rep=31, majority_threshold=50)
else:
    print(f"{senate_state_csv} not found. Run US_Senate_results.py first.")
```

```python
# Senate County-Level Inspection & Turnout Projections
if os.path.exists(senate_county_csv):
    df_senate_county = pd.read_csv(senate_county_csv)
    print(f"Loaded {len(df_senate_county)} county/town records across {df_senate_county['State'].nunique()} states.")
    display(df_senate_county.head(10))
    plot_turnout_projections(df_senate_county, turnout_rate=0.60)
else:
    print(f"{senate_county_csv} not found.")
```

## 2. U.S. House of Representatives Analysis (2026)

- Total House Seats: 435 (218 needed for majority)
- All 435 congressional districts are up for election in 2026.

```python
house_csv = 'US_House_results.csv'

if os.path.exists(house_csv):
    df_house = pd.read_csv(house_csv)
    display(df_house.head(10))
    plot_party_seats(df_house, majority_threshold=218)
else:
    print(f"{house_csv} not found. Run US_House_results.py first.")
```
